### 1. **What is Generative AI?**
   Generative AI refers to AI systems that can generate new content (text, images, audio, etc.) that resembles human-created data, based on patterns learned from training data.

### 2. **How is Generative AI different from traditional AI?**
   Traditional AI focuses on predictions and classifications, while Generative AI creates new data, such as writing stories or generating images.

### 3. **Name two applications of Generative AI in the industry:**

   * Content generation (e.g., copywriting, image generation)
   * Drug discovery (e.g., molecule design)

### 4. **What are some challenges associated with Generative AI?**

   * Data quality and bias
   * Ethical concerns (e.g., deepfakes)
   * Computational cost

### 5. **Why is Generative AI important for modern applications?**
   It enables automation in creative tasks, improves personalization, and drives innovation across various sectors.

### 6. **What is probabilistic modeling in the context of Generative AI?**
   It involves using probability distributions to model the uncertainty and variability in data generation.

### 7. **Define a generative model:**
   A model that learns the joint probability distribution of input features and outputs to generate new data.

### 8. **Explain how an n-gram model works in text generation:**
   It predicts the next word based on the previous n-1 words using probabilities derived from a text corpus.

### 9. **What are the limitations of n-gram models?**

   * Limited context window
   * Data sparsity
   * Poor performance on rare word sequences

### 10. **How can you improve the performance of an n-gram model?**

    * Smoothing techniques
    * Using backoff models
    * Larger n-gram sizes

### 11. **What is the Markov assumption, and how does it apply to text generation?**
    It assumes that the probability of a word depends only on a limited number of preceding words, not the entire sequence.

### 12. **Why are probabilistic models important in generative AI?**
    They provide a principled way to handle uncertainty and generate diverse outputs.

### 13. **What is an autoencoder?**
    A neural network that learns to compress (encode) data and then reconstruct (decode) it, typically used for unsupervised learning.

### 14. **How does a VAE differ from a standard autoencoder?**
    A VAE introduces a probabilistic latent space and regularization using KL divergence to allow controlled generation.

### 15. **Why are VAEs useful in generative modeling?**
    They generate diverse and coherent outputs due to the structured latent space.

### 16. **What role does the decoder play in an autoencoder?**
    It reconstructs the input data from the compressed latent representation.

### 17. **How does the latent space affect text generation in a VAE?**
    It determines the diversity and similarity of generated samples by sampling from a continuous space.

### 18. **What is the purpose of the Kullback-Leibler (KL) divergence term in VAEs?**
    It regularizes the latent space to follow a standard distribution, ensuring smoothness and generalization.

### 19. **How can you prevent overfitting in a VAE?**

    * Using dropout
    * Early stopping
    * Limiting latent space dimensions

### 20. **Explain why VAEs are commonly used for unsupervised learning tasks:**
    They learn representations without labeled data and can generate new samples.

### 21. **What is a transformer model?**
    A neural network architecture that uses self-attention mechanisms to process sequential data in parallel.

### 22. **Explain the purpose of self-attention in transformers:**
    It allows the model to weigh the importance of different words in a sequence for better context understanding.

### 23. **How does a GPT model generate text?**
    It predicts the next token based on previous tokens using an autoregressive decoder-only transformer.

### 24. **What are the key differences between a GPT model and an RNN?**

    * GPT uses self-attention, not recurrence
    * GPT processes data in parallel during training
    * GPT handles long-range dependencies better
### 25. **How does fine-tuning improve a pre-trained GPT model?**
    It adapts the model to a specific task or domain using additional labeled data.

### 26. **What is zero-shot learning in the context of GPT models?**
    The ability to perform tasks without explicit training, based solely on instructions given in prompts.

### 27. **Describe how prompt engineering can impact GPT model performance:**
    Well-designed prompts guide the model to produce more accurate, relevant, and task-specific responses.

### 28. **Why are large datasets essential for training GPT models?**
    They help the model learn diverse language patterns, improving generalization and fluency.

### 29. **What are potential ethical concerns with GPT models?**

    * Misinformation generation
    * Bias amplification
    * Privacy violations

### 30. **How does the attention mechanism contribute to GPT’s ability to handle long-range dependencies?**
    It allows each token to attend to all other tokens in a sequence, capturing global context.

### 31. **What are some limitations of GPT models for real-world applications?**

    * High computational requirements
    * Hallucinations (false information)
    * Sensitivity to prompt wording

### 32. **How can GPT models be adapted for domain-specific text generation?**

    * Fine-tuning on domain-specific data
    * Using prompt engineering with relevant examples

### 33. **What are some common metrics for evaluating text generation quality?**

    * BLEU
    * ROUGE
    * Perplexity

### 34. **Explain the difference between deterministic and probabilistic text generation:**
    Deterministic (e.g., greedy decoding) always chooses the highest probability word; probabilistic (e.g., sampling) adds randomness.

### 35. **How does beam search improve text generation in language models?**
    It explores multiple sequences at each step, balancing quality and diversity for better outputs.

```python
import random

text = "The cat is on the mat"
words = text.split()
markov_chain = {}

for i in range(len(words) - 1):
    word, next_word = words[i], words[i + 1]
    if word not in markov_chain:
        markov_chain[word] = []
    markov_chain[word].append(next_word)

# Generate a sentence
word = "The"
sentence = [word]
for _ in range(5):
    next_words = markov_chain.get(word, [])
    if not next_words:
        break
    word = random.choice(next_words)
    sentence.append(word)

print(" ".join(sentence))
```

```python
from keras.models import Model
from keras.layers import Input, LSTM, RepeatVector
from keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
import numpy as np

sentences = ["The cat is on the mat", "The dog is in the fog", "Birds fly in the sky"]
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
vocab_size = len(tokenizer.word_index) + 1

seqs = tokenizer.texts_to_sequences(sentences)
max_len = max(len(seq) for seq in seqs)
padded = pad_sequences(seqs, maxlen=max_len)

input_dim = vocab_size
timesteps = max_len
embed_dim = 8
latent_dim = 16

inputs = Input(shape=(timesteps,))
x = LSTM(latent_dim)(inputs)
encoded = RepeatVector(timesteps)(x)
decoded = LSTM(input_dim, return_sequences=True)(encoded)

autoencoder = Model(inputs, decoded)
autoencoder.compile(optimizer='adam', loss='sparse_categorical_crossentropy')
# You would need to reshape the padded input to match the target shape for training
```

```python
from keras.models import Sequential
from keras.layers import SimpleRNN, Dense, Embedding
import numpy as np

# Example dummy dataset
chars = "abcdefghijklmnopqrstuvwxyz "
char_to_idx = {c: i for i, c in enumerate(chars)}
idx_to_char = {i: c for c, i in char_to_idx.items()}

seq_length = 5
text = "hello world hello world"
X, y = [], []
for i in range(len(text) - seq_length):
    seq = text[i:i+seq_length]
    label = text[i+seq_length]
    X.append([char_to_idx[c] for c in seq])
    y.append(char_to_idx[label])

X = np.array(X)
y = np.array(y)

model = Sequential()
model.add(Embedding(input_dim=len(chars), output_dim=10, input_length=seq_length))
model.add(SimpleRNN(64))
model.add(Dense(len(chars), activation='softmax'))
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
model.fit(X, y, epochs=50)
```

```python
import numpy as np
from keras.models import Sequential
from keras.layers import LSTM, Dense, Embedding

# --- Prepare data ---
text = "hello world. this is ajit from bihar learning text generation using lstm."
chars = sorted(list(set(text)))
char_to_idx = {c: i for i, c in enumerate(chars)}
idx_to_char = {i: c for i, c in enumerate(chars)}
vocab_size = len(chars)
seq_length = 10
step = 1

sequences = []
next_chars = []
for i in range(0, len(text) - seq_length, step):
    sequences.append(text[i: i + seq_length])
    next_chars.append(text[i + seq_length])

X = np.zeros((len(sequences), seq_length), dtype=np.int32)
y = np.zeros((len(sequences), vocab_size), dtype=np.bool_)

for i, seq in enumerate(sequences):
    X[i] = [char_to_idx[char] for char in seq]
    y[i][char_to_idx[next_chars[i]]] = 1

# --- Build model ---
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=50, input_length=seq_length),
    LSTM(128),
    Dense(vocab_size, activation='softmax')
])
model.compile(loss='categorical_crossentropy', optimizer='adam')
model.summary()

# --- Train model ---
model.fit(X, y, batch_size=64, epochs=100, verbose=2)

# --- Text generation function ---
def generate_text(seed, length=100):
    generated = seed
    for _ in range(length):
        input_seq = [char_to_idx.get(char, 0) for char in generated[-seq_length:]]
        input_seq = np.reshape(input_seq, (1, seq_length))
        prediction = model.predict(input_seq, verbose=0)
        next_char = idx_to_char[np.argmax(prediction)]
        generated += next_char
    return generated

# --- Generate and print text ---
print("Generated text:")
print(generate_text("hello worl"))
```
