## Attention based Models and Transfer Learning

### 1. What is BERT and how does it work?
BERT (Bidirectional Encoder Representations from Transformers) is a transformer-based model pre-trained on large text corpora using masked language modeling and next sentence prediction. It processes text bidirectionally, meaning it looks at both left and right context simultaneously to understand word meaning.

###2. What are the main advantages of using the attention mechanism in neural networks?
Handles long-range dependencies better than RNNs.

Focuses on important parts of input dynamically.

Parallelizable, improving training speed.

Reduces reliance on sequence order.

### 3. How does the self-attention mechanism differ from traditional attention mechanisms?
Self-attention relates each word in a sequence to every other word in the same sequence.

Traditional attention often involves attention from a decoder to an encoder (e.g., in Seq2Seq models).

### 4. What is the role of the decoder in a Seq2Seq model?
The decoder generates the output sequence, step-by-step, using the encoded input context and previous outputs.

### 5. What is the difference between GPT-2 and BERT models?
Aspect	BERT	GPT-2
Directionality	Bidirectional	Unidirectional (left to right)
Pretraining	MLM + NSP	Language modeling
Usage	Classification & QA	Text generation

### 6. Why is the Transformer model considered more efficient than RNNs and LSTMs?
Parallel processing (not sequential like RNNs)

Captures global dependencies via attention

Scales better with large datasets

### 7. Explain how the attention mechanism works in a Transformer model.
Each word’s representation is updated by computing a weighted sum of all words in the sequence using three vectors:

Query (Q)

Key (K)

Value (V)
The weights come from dot products of Q and K, followed by a softmax.

### 8. What is the difference between an encoder and a decoder in a Seq2Seq model?
Encoder: Converts input sequence into context vector.

Decoder: Uses context + past outputs to generate the next word in the output sequence.

### 9. What is the primary purpose of using the self-attention mechanism in transformers?
To allow the model to dynamically weigh the importance of each word relative to others in a sequence, enabling understanding of context and relationships.

### 10. How does the GPT-2 model generate text?
GPT-2 generates text in an autoregressive manner, predicting the next word based on previous ones using a stack of transformer decoder blocks.

### 11. What is the main difference between the encoder-decoder architecture and a simple neural network?
Encoder-decoder can handle sequences of varying lengths.

Simple neural networks lack memory and struggle with sequential context.

### 12. Explain the concept of “fine-tuning” in BERT.
Fine-tuning is training the pretrained BERT model on a specific task (e.g., sentiment analysis) with labeled data by updating its weights slightly.

### 13. How does the attention mechanism handle long-range dependencies in sequences?
It enables any word to attend to any other word, regardless of distance, unlike RNNs which struggle with far-apart tokens.

### 14. What is the core principle behind the Transformer architecture?
Replaces recurrence with self-attention and feedforward layers, enabling global context modeling and efficient parallel training.

### 15. What is the role of the "position encoding" in a Transformer model?
Since Transformers don’t have sequence order by design, position encodings add order information to input embeddings using sinusoidal or learned patterns.

### 16. How do Transformers use multiple layers of attention?
Each layer consists of multi-head attention + feedforward networks. Stacking layers allows learning more abstract relationships in data.

### 17. What does it mean when a model is described as “autoregressive” like GPT-2?
It predicts the next word based on previous words only (left-to-right), generating text sequentially.

### 18. How does BERT's bidirectional training improve its performance?
By looking at both sides of a word, BERT captures richer context, which improves accuracy on tasks like Q&A and classification.

### 19. What are the advantages of using the Transformer over RNN-based models in NLP?
Faster training (parallelism)

Better context modeling

Handles long-range dependencies

Scalable and more effective for large datasets

### 20. What is the attention mechanism’s impact on the performance of models like BERT and GPT-2?
It helps these models:

Focus on relevant context

Understand relationships in text

Achieve state-of-the-art accuracy on many NLP tasks

### 1. How to implement a simple text classification model using LSTM in Keras0


```python
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense
from keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences

# Example data
texts = ['I love this movie', 'This film was terrible']
labels = [1, 0]  # 1: Positive, 0: Negative

# Tokenization and padding
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
X = pad_sequences(sequences, maxlen=10)

# Model
model = Sequential()
model.add(Embedding(input_dim=1000, output_dim=64, input_length=10))
model.add(LSTM(64))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X, labels, epochs=5)
```

### 2. How to generate sequences of text using a Recurrent Neural Network (RNN)

```python
# Assume trained model and tokenizer
seed_text = "Once upon a"
for _ in range(20):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=10, padding='pre')
    predicted = model.predict(token_list, verbose=0)
    next_index = predicted.argmax()
    next_word = tokenizer.index_word[next_index]
    seed_text += " " + next_word

print(seed_text)
```

### 5.How to implement a simple Seq2Seq model for machine translation using LSTM in Keras0

```python
from keras.models import Model
from keras.layers import Input, LSTM, Dense

# Encoder
encoder_inputs = Input(shape=(None,))
enc_emb = Embedding(input_dim=10000, output_dim=256)(encoder_inputs)
encoder_lstm, state_h, state_c = LSTM(256, return_state=True)(enc_emb)
encoder_states = [state_h, state_c]

# Decoder
decoder_inputs = Input(shape=(None,))
dec_emb = Embedding(input_dim=10000, output_dim=256)(decoder_inputs)
decoder_lstm = LSTM(256, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(dec_emb, initial_state=encoder_states)
decoder_dense = Dense(10000, activation='softmax')
decoder_outputs = decoder_dense(decoder_outputs)

model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
model.compile(optimizer='rmsprop', loss='categorical_crossentropy')
```

### 6.How to generate text using a pre-trained transformer model (GPT-2)

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

input_ids = tokenizer.encode("The future of AI is", return_tensors='pt')
output = model.generate(input_ids, max_length=50, num_return_sequences=1)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```
